1. What is a RESTful API
Ans 1. A RESTful API (Representational State Transfer API) is a design approach for web services that treats server-side data as resources, each identified by a unique URL and manipulated using standard HTTP methods. In a RESTful system, clients perform operations—such as retrieving a resource with GET, creating one with POST, updating it with PUT or PATCH, and deleting it with DELETE—while the server remains stateless, meaning each request contains all the information needed to process it. Resources are often represented in formats like JSON or XML, making them easy to read and parse across different platforms. By adhering to REST principles—uniform interface, statelessness, cacheability, client-server separation, layered system, and optional code on demand—RESTful APIs achieve simplicity, scalability, and flexibility, allowing diverse clients (web browsers, mobile apps, IoT devices) to interact seamlessly with the same back-end services.


2. Explain the concept of API specification
Ans 2. An API specification is a formal, machine‐ and human‐readable contract that defines how different software components communicate with each other. It describes the available endpoints (URLs or routes), the HTTP methods (GET, POST, PUT, DELETE, etc.) each endpoint accepts, the structure and data types of request parameters and response bodies, authentication requirements, error codes, and any other protocols or conventions that clients must follow. By codifying these details—often in a standardized format like OpenAPI (formerly Swagger) or RAML—an API specification ensures consistency, enables automatic generation of client libraries and documentation, facilitates testing and validation, and reduces integration errors. In essence, it serves as a blueprint that both the API provider and its consumers rely on to build, test, and maintain interoperable systems.

4. What is routing in Flask
Ans 4. In Flask, routing refers to the mechanism by which you map incoming HTTP request URLs to the specific Python functions (called view functions) that should handle them. When you decorate a function with `@app.route('/path')`, you’re telling Flask, “If a client visits this URL path, invoke this function and return its result as the HTTP response.” Routes can include variable components (e.g., `/user/<int:id>`) so that dynamic values from the URL are passed into your view function as arguments. Flask’s routing system also lets you specify which HTTP methods (GET, POST, etc.) a particular endpoint should accept. Internally, Flask maintains a URL map, and every time a request arrives, it searches that map for the best-matching route, invokes the associated function, and sends back whatever that function returns—be it HTML, JSON, or a redirect. This clear separation of URL patterns and view logic makes it easy to build organized, RESTful web applications.

6. What are HTTP methods used in RESTful APIs
Ans 6. In RESTful APIs, a small set of standard HTTP methods is used to perform Create, Read, Update, and Delete (CRUD) operations on resources, each method carrying a clear, semantic meaning. **GET** retrieves a representation of a resource without altering it, making it safe and idempotent, while **POST** submits new data to create a subordinate resource and is neither safe nor idempotent. **PUT** uploads a complete representation of a resource to a specified URI—overwriting it if it exists or creating it otherwise—and is idempotent, meaning repeated identical requests produce the same result. **PATCH** applies partial modifications to an existing resource and is not strictly idempotent unless the patch itself is designed that way. **DELETE** removes the resource identified by the URI and is idempotent: once deleted, further delete requests have no additional effect. **HEAD** behaves like GET but returns only the response headers, allowing clients to check metadata such as content length or last-modified date without transferring the entire payload. Finally, **OPTIONS** lets clients discover the supported HTTP methods and other communication options for a resource, facilitating cross-origin requests and dynamic client interfaces. By leveraging these methods consistently, RESTful services maintain a uniform interface and predictable behavior across different resources.


7. What is the purpose of the @app.route() decorator in Flask
Ans 7. The `@app.route()` decorator in Flask serves as the primary mechanism for URL routing: it tells the Flask application which URL endpoint should trigger execution of the decorated view function. When you write `@app.route('/hello')` above a function, Flask registers that function to handle incoming HTTP requests to the `/hello` path, automatically invoking it and returning its response whenever a client navigates to that URL. You can also specify accepted HTTP methods (e.g., `methods=['GET', 'POST']`), URL parameters (`/user/<username>`), and even multiple routes for a single function. Under the hood, Flask builds a routing map from these decorators at startup, matching each request’s path and method to the appropriate handler. This declarative approach keeps your code clear and concise, cleanly associating each URL with its processing logic.


8. What is the difference between GET and POST HTTP methods
Ans 8. The primary difference between the GET and POST HTTP methods lies in how they send data to the server and how they’re intended to be used. **GET** requests include all data in the URL’s query string (e.g., `?user=alice&age=30`), making them ideal for retrieving resources or performing idempotent operations—actions that can be repeated without changing the server’s state. Because GET parameters are visible in the URL, they’re limited in length and less secure for sensitive information. Additionally, GET requests are cacheable by default and can be bookmarked or shared easily. In contrast, **POST** requests encapsulate data within the request body rather than the URL, allowing for larger payloads (such as file uploads or complex JSON objects) and keeping parameters out of the browser history. POST is intended for operations that change server state—creating, updating, or deleting resources—and is not idempotent (repeating a POST may result in multiple creations). POST requests are not cached by default and cannot be bookmarked, which makes them better suited for submitting forms or performing transactions where data privacy and side effects matter.


9. How do you handle errors in Flask APIs
Ans 9. In Flask APIs, I handle errors by defining centralized error handlers that catch both HTTP exceptions and unexpected server errors, ensuring that every response follows a consistent JSON structure. For anticipated errors—such as invalid input, missing resources, or unauthorized access—I raise `werkzeug.exceptions.HTTPException` (e.g., via `abort(404, description="Item not found")`) and implement `@app.errorhandler()` decorators to intercept these exceptions, format a clear message, and return the appropriate status code. For unanticipated exceptions, I register a generic `@app.errorhandler(Exception)` handler that logs the full traceback for debugging but returns a sanitized JSON response with a 500 status code to avoid leaking internal details. Throughout, I include helpful error codes and human-readable messages in the response body, integrate input validation (using libraries like Marshmallow or Pydantic) to catch errors early, and ensure that all endpoints consistently return well-structured error objects, which makes it easier for clients to programmatically handle and display errors.


10.  How do you connect Flask to a SQL database

This setup seamlessly connects your Flask application to your chosen SQL database, managing connections, transactions, and schema definitions under the hood.

12. What are Flask blueprints, and how are they useful
ANs 12. Flask blueprints are a way to organize a Flask application into modular, reusable components by grouping related routes, templates, static files, and error handlers into distinct “mini-applications.” Instead of defining all your view functions and configurations in a single file or app object, you create one or more blueprints—each encapsulating a particular feature or set of endpoints (for example, an authentication module, an admin dashboard, or an API). This separation of concerns makes your codebase cleaner and more maintainable, allows you to develop and test each part in isolation, and even lets you distribute or reuse entire blueprints across different projects with minimal effort.


13. What is the purpose of Flask's request object
Ans 13. In Flask, the `request` object serves as a convenient, thread-local proxy to all the data sent by the client in an HTTP request. By importing `from flask import request`, you gain access to properties and methods that let you inspect query parameters (`request.args`), form data (`request.form`), JSON payloads (`request.get_json()`), uploaded files (`request.files`), headers (`request.headers`), cookies (`request.cookies`), and even metadata like the request method (`request.method`) and URL (`request.url`). Behind the scenes, Flask uses Werkzeug’s request parsing machinery to turn raw HTTP into Python objects, so you can write clean, readable code for handling input without worrying about low-level socket or protocol details. This unified interface makes it easy to validate, sanitize, and respond to client data in a secure and structured way.


14. How do you create a RESTful API endpoint using Flask
Ans 14. To create a RESTful API endpoint in Flask, you first install and import the Flask framework, then initialize your application with `app = Flask(__name__)`. Finally, you run the application (often with `app.run()`) or configure it under a WSGI server for production. This simple pattern—route declaration, method handling, request parsing, processing, and JSON response—forms the backbone of every RESTful API endpoint in Flask.


15. What is the purpose of Flask's jsonify() function

Internally, it serializes the passed data to a JSON‐encoded string, sets the `Content-Type` header to `application/json`, and wraps the result in a Flask `Response` object. 16. Explain Flask’s url_for() function
Ans 16. In Flask, the `url_for()` function is a powerful utility that dynamically generates URLs for your application’s routes and static assets, ensuring that links remain valid even if you later refactor your view functions or change your URL patterns. Instead of hard-coding paths, you call `url_for('endpoint_name', **values)`—where `'endpoint_name'` is the name of the view function (or `'static'` for static files)—and Flask will look up the corresponding URL, inserting any variable parts (like `<int:id>`) from the `values` you pass. This not only keeps your templates and redirects DRY (Don’t Repeat Yourself), but also guards against errors from typos or forgotten updates: if you rename an endpoint or adjust its route, every `url_for()` call automatically reflects the change. Additionally, `url_for()` handles URL quoting and can generate absolute URLs when needed, making it an essential tool for building robust, maintainable Flask applications.

19.  What are HTTP status codes, and why are they important in a Flask API
Ans 19. HTTP status codes are standardized three-digit numbers returned by a server in response to a client's request, indicating the outcome of that request. In the context of a Flask API, these codes are crucial because they provide clear, consistent communication between the server and client about what happened during the request-processing cycle. For example, a status code of **200** means the request was successful, **201** indicates a new resource was created, **400** signals a bad request due to client-side error, and **500** represents a server-side error. These codes help developers and users understand whether a request was processed correctly or if an error occurred, and what type of error it was. Proper use of HTTP status codes in a Flask API improves the API's usability, debuggability, and compliance with RESTful design principles. It ensures that clients (such as frontend applications or other services) can programmatically handle different scenarios based on the response status, making the API more robust and reliable.


20. How do you handle POST requests in Flask
Ans 20. In Flask, handling **POST requests** is straightforward and commonly done using the `@app.route()` decorator with the `methods` parameter set to include `'POST'`. Typically, this is used in forms or APIs where data is sent from the client to the server. Inside the route function, you can access the data sent in the POST request using the `request` object from Flask’s `flask` module. For example, form data submitted via an HTML form can be retrieved using `request.form`, while JSON payloads are accessed using `request.get_json()`. Here's a basic example:

```python
from flask import Flask, request

app = Flask(__name__)

@app.route('/submit', methods=['POST'])
def handle_post():
    name = request.form['name']  # for form data
    # Or for JSON data: name = request.get_json().get('name')
    return f"Received data for {name}"
```

In this example, when a POST request is made to the `/submit` route, the server retrieves the submitted data and returns a simple response. It’s also important to handle potential errors and validate the data before processing it. POST requests are commonly used when building APIs or handling sensitive input, such as login forms or data submission, where information needs to be securely transmitted from client to server.


21. How would you secure a Flask API
Ans 21. To secure a Flask API, you should first serve it over HTTPS to encrypt all data in transit, then enforce robust authentication and authorization—commonly via API keys or JSON Web Tokens (JWTs) managed by extensions like Flask-JWT-Extended or OAuth2 workflows. Validate and sanitize every incoming request to prevent SQL injection or malicious payloads, using parameterized queries and libraries such as Marshmallow or Cerberus for schema enforcement. Implement rate limiting (e.g., with Flask-Limiter) and IP blacklisting to guard against denial-of-service and brute-force attacks. Configure strict CORS policies to control which origins can access your endpoints, and enable CSRF protection if you expose any session-based routes. Finally, maintain detailed logging of authentication attempts and errors, regularly update dependencies to patch vulnerabilities, and consider adding a Web Application Firewall (WAF) or API gateway for an additional layer of threat filtering.

23.What is the role of Flask’s session object
Ans 23. In Flask, the `session` object plays a crucial role in maintaining **user-specific data across multiple requests** during a web session. Since HTTP is a stateless protocol, Flask's session mechanism provides a way to store data (like user preferences, authentication status, or shopping cart contents) between requests from the same user. The `session` object works like a Python dictionary and stores information that you want to persist throughout a user's interaction with the web application. By default, Flask stores session data on the client-side using a secure, encrypted cookie, signed with the app’s secret key to ensure data integrity and prevent tampering. This makes sessions lightweight and fast, as no server-side storage is required unless explicitly implemented. However, because the session data is stored in cookies, it’s important to avoid storing sensitive information or large amounts of data. Overall, the `session` object is a simple yet powerful tool in Flask to manage temporary user data and enable features like login persistence and user-specific behavior.

````python
#6. How do you handle forms in Flask
!pip install flask-ngrok
!pip install pyngrok
!ngrok authtoken '2xFkON9cZKJ5L1jgeQMEIYRHCGJ_2pz8dbd6xqMoPN35VtH9X'
from flask import Flask, request, render_template_string, redirect, url_for
from flask_ngrok import run_with_ngrok

app = Flask(__name__)
run_with_ngrok(app)  # Automatically starts ngrok when you run the app

# HTML template
form_html = '''
<!doctype html>
<title>Flask Form Example</title>
<h2>Enter Your Details</h2>
<form method=post>
  Name: <input type=text name=name><br><br>
  Email: <input type=email name=email><br><br>
  <input type=submit value=Submit>
</form>
{% if name %}
  <h3>Hello {{ name }}! Your email is {{ email }}.</h3>
{% endif %}
'''

@app.route('/', methods=['GET', 'POST'])
def form_handler():
    if request.method == 'POST':
        name = request.form.get('name')
        email = request.form.get('email')
        return render_template_string(form_html, name=name, email=email)
    return render_template_string(form_html)

if __name__ == '__main__':
    app.run()
````

````python
#7. How can you validate form data in Flask
!pip install flask-ngrok
!pip install pyngrok
!ngrok authtoken '2xFkON9cZKJ5L1jgeQMEIYRHCGJ_2pz8dbd6xqMoPN35VtH9X'
from flask import Flask, render_template_string, request, flash, redirect, url_for
from flask_wtf import FlaskForm
from wtforms import StringField
from wtforms.validators import DataRequired, Email
from pyngrok import ngrok
import os

app = Flask(__name__)
app.secret_key = os.urandom(24)  # Required for session handling

# WTForms form class
class MyForm(FlaskForm):
    name = StringField('Name', validators=[DataRequired()])
    email = StringField('Email', validators=[DataRequired(), Email()])

# Home route with form
@app.route('/', methods=['GET', 'POST'])
def index():
    form = MyForm()
    if form.validate_on_submit():
        name = form.name.data
        email = form.email.data
        flash(f'Form submitted successfully with Name: {name}, Email: {email}', 'success')
        return redirect(url_for('index'))
    return render_template_string('''
    <!doctype html>
    <title>Form Validation</title>
    <h1>Flask Form</h1>
    <form method="POST">
        {{ form.hidden_tag() }}
        <p>
            {{ form.name.label }}<br>
            {{ form.name(size=32) }}<br>
            {% for error in form.name.errors %}
                <span style="color: red;">[{{ error }}]</span>
            {% endfor %}
        </p>
        <p>
            {{ form.email.label }}<br>
            {{ form.email(size=32) }}<br>
            {% for error in form.email.errors %}
                <span style="color: red;">[{{ error }}]</span>
            {% endfor %}
        </p>
        <p><input type="submit" value="Submit"></p>
    </form>
    {% with messages = get_flashed_messages(with_categories=true) %}
        {% if messages %}
            <ul>
            {% for category, message in messages %}
                <li style="color: green;">{{ message }}</li>
            {% endfor %}
            </ul>
        {% endif %}
    {% endwith %}
    ''', form=form)

# Main with optional ngrok tunnel
if __name__ == '__main__':
    use_ngrok = True  # Change to False to disable ngrok
    port = 5000

    if use_ngrok:
        public_url = ngrok.connect(port)
        print(f' * ngrok tunnel running at: {public_url}')

    app.run(port=port)
````

````python
#8. How do you manage sessions in Flask
!pip install flask-ngrok
!pip install pyngrok
!ngrok authtoken '2xFkON9cZKJ5L1jgeQMEIYRHCGJ_2pz8dbd6xqMoPN35VtH9X'
from flask import Flask, session, redirect, url_for, request, render_template_string

app = Flask(__name__)
app.secret_key = 'your_secret_key'  # Required to use sessions securely

# Home page
@app.route('/')
def home():
    if 'username' in session:
        return f"Logged in as {session['username']} <br><a href='/logout'>Logout</a>"
    return "You are not logged in <br><a href='/login'>Login</a>"

# Login page
@app.route('/login', methods=['GET', 'POST'])
def login():
    if request.method == 'POST':
        session['username'] = request.form['username']
        return redirect(url_for('home'))

    # Simple HTML form
    return render_template_string('''
        <form method="post">
            <p><input type=text name=username>
            <p><input type=submit value=Login>
        </form>
    ''')

# Logout route
@app.route('/logout')
def logout():
    session.pop('username', None)  # Remove username from session
    return redirect(url_for('home'))

if __name__ == '__main__':
    app.run(debug=True)
````

````python
#11. How do you structure a Flask app using Blueprints
!pip install flask-ngrok
!pip install pyngrok
!ngrok authtoken '2xFkON9cZKJ5L1jgeQMEIYRHCGJ_2pz8dbd6xqMoPN35VtH9X'
from flask import Blueprint, render_template, redirect, url_for

auth_bp = Blueprint('auth', __name__, url_prefix='/auth')

@auth_bp.route('/login')
def login():
    return render_template('auth/login.html')

@auth_bp.route('/logout')
def logout():
    return redirect(url_for('auth.login'))
from flask import Blueprint, render_template

blog_bp = Blueprint('blog', __name__, url_prefix='/blog')

@blog_bp.route('/')
def index():
    return render_template('blog/post.html')

@blog_bp.route('/post/<int:post_id>')
def post(post_id):
    return f"Showing post {post_id}"
from flask import Flask
from blueprints.auth import auth_bp
from blueprints.blog import blog_bp

def create_app():
    app = Flask(__name__)

    # Register Blueprints
    app.register_blueprint(auth_bp)
    app.register_blueprint(blog_bp)

    @app.route('/')
    def home():
        return "Welcome to the Flask app with Blueprints!"

    return app

if __name__ == '__main__':
    app = create_app()
    app.run(debug=True)
````

````python
#12. How do you define a custom Jinja filter in Flask
!pip install flask-ngrok
!pip install pyngrok
!ngrok authtoken '2xFkON9cZKJ5L1jgeQMEIYRHCGJ_2pz8dbd6xqMoPN35VtH9X'
from flask import Flask, render_template_string

app = Flask(__name__)

# Step 1: Define your custom filter function
def reverse_string(s):
    return s[::-1]

# Step 2: Register the filter with Flask's Jinja environment
app.jinja_env.filters['reverse'] = reverse_string

# Optional: Use a route to test it
@app.route('/')
def index():
    text = "Flask is awesome!"
    # Using render_template_string for demo purposes
    return render_template_string("Original: {{ text }} <br> Reversed: {{ text|reverse }}", text=text)

if __name__ == '__main__':
    app.run(debug=True)
````

````python
#13.  How can you redirect with query parameters in Flask
!pip install flask-ngrok
!pip install pyngrok
!ngrok authtoken '2xFkON9cZKJ5L1jgeQMEIYRHCGJ_2pz8dbd6xqMoPN35VtH9X'
from flask import Flask, redirect, url_for, request

app = Flask(__name__)

# Initial route that redirects with query parameters
@app.route('/go-to-result')
def go_to_result():
    return redirect(url_for('result', name='Mustahid', age=25))

# Target route that receives query parameters
@app.route('/result')
def result():
    name = request.args.get('name')
    age = request.args.get('age')
    return f'Name: {name}, Age: {age}'

if __name__ == '__main__':
    app.run(debug=True)
````
